<h2> check branches<h2>

```python
display(
    spark.sql("""
        SELECT
            _source_branch,
            COUNT(*) AS record_count
        FROM hospital_catalog.bronze.patients
        GROUP BY _source_branch
        ORDER BY _source_branch
    """)
)
```

<h2>Check source systems<h2>

```python
display(
    spark.sql("""
        SELECT
            _source_file,
            COUNT(*) AS record_count
        FROM hospital_catalog.bronze.patients
        GROUP BY _source_file
        ORDER BY _source_file
    """)
)
```

<h2>Check ingestion batches<h2>

```python
display(
    spark.sql("""
        SELECT
            _batch_id,
            _source_branch,
            COUNT(*) AS record_count
        FROM hospital_catalog.bronze.patients
        GROUP BY
            _batch_id,
            _source_branch
        ORDER BY _batch_id
    """)
)
```

<h2>Check for null source metadata<h2>

```python
display(
    spark.sql("""
        SELECT
            COUNT(*) AS total_rows,
            SUM(
                CASE
                    WHEN _source_branch IS NULL
                    THEN 1
                    ELSE 0
                END
            ) AS missing_branch,
            SUM(
                CASE
                    WHEN _source_system IS NULL
                    THEN 1
                    ELSE 0
                END
            ) AS missing_source_system,
            SUM(
                CASE
                    WHEN _source_file IS NULL
                    THEN 1
                    ELSE 0
                END
            ) AS missing_source_file
        FROM hospital_catalog.bronze.patients
    """)
)
```
