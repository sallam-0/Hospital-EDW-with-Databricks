```python
%sql
-- Breakdown of rejections by rule and severity
SELECT 
    _quarantine_rule_id,
    _quarantine_severity,
    _quarantine_reason,
    COUNT(*) AS rejected_count
FROM hospital_catalog.quarantine.patients
GROUP BY _quarantine_rule_id, _quarantine_severity, _quarantine_reason
ORDER BY rejected_count DESC;
```
